# Assemble ttl labels, times and add-on data (csv file)

### Imports

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

### Functions

## Define paths and parameters to access data

```python
recording = 'M296_3'
ttl_type = 'obx_stim_pulses' # obx_stim_pulses/obx_plbs
csv_path = rf'M:\test_Stim_g0\test_Stim_g0_imec0\catgt_test_Stim_g0' # csv file and txt of ttl times should be in same folder
csv_name = None
ttl_file = 'M296_3_obx_stim_pulses_seconds.txt'  # Add txt file with onset event times
```

**Use in case of stim_pulses or absence of csv file with events data**

Write number of events per label, if a label is repeated after an event with another label [a, b, a],
then write again label with corresponding number. Ex. {'tr100ms':23, 'tr1s':16, 'tr100ms':15}

```python
stim_dict = {'tr100ms':23, 'tr1s':16} # Doesnt do anything, only if csv_name is None
```

## Load data

```python
if csv_name is not None:
    csv_full_path = Path(csv_path, csv_name)
    events_df = pd.read_csv(csv_full_path)
else:
    # Repeat each key based on its integer value
    event_labels = [key for key, count in stim_dict.items() for _ in range(count)]
    events_df = pd.DataFrame({'s_names': event_labels})

# Add ttl_times to events_df
ttl_full_path = Path(csv_path, ttl_file)
ttl_times = np.loadtxt(ttl_full_path) # ttl onset times
events_df.insert(1, 's_times', ttl_times)

# Add ttl_TPrime times to events_df
ttl_tp_full_path = Path(csv_path, ttl_file.replace('.txt', '_tprime.txt'))
ttl_tp_times = np.loadtxt(ttl_tp_full_path) # ttl TPrime onset times
events_df.insert(2, 's_timesP', ttl_tp_times)
```

## Save events_df as csv file

```python
save_full_path = Path(csv_path, f'{recording}_{ttl_type}_exp_summary.csv')
events_df.to_csv(save_full_path, index=False)
```
